```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.metrics import accuracy_score

#  Load the dataset
data = pd.read_csv('newly created pesticides sheet.csv')  # Replace 'your_file.csv' with your file name

#  Clean data
data = data[pd.to_numeric(data['Soil pH'], errors='coerce').notnull()]  # Fix Soil pH issues
data.reset_index(drop=True, inplace=True)

#  Encode categorical columns
categorical_columns = ['State', 'Crop', 'Crop Stage', 'Soil Type', 'Season',
                       'Past Fertilizer Used', 'Past Pesticide Used',
                       'Farmer Preference', 'Pesticide Type', 'Pest/Symptom Observed', 
                       'Pest Severity', 'Application Method']  
label_encoders = {}
for col in categorical_columns:
    le = LabelEncoder()
    data[col] = le.fit_transform(data[col])
    label_encoders[col] = le

# Encode Target Columns
y_pesticide = LabelEncoder().fit_transform(data['Recommended Pesticide'])
y_spraytime = LabelEncoder().fit_transform(data['Best Spray Time'])

#  Features and Target
X = data.drop(['Recommended Pesticide', 'Best Spray Time'], axis=1)

#  Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y_pesticide, test_size=0.2, random_state=42)

#  Feature Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Build the MLP model
model = Sequential()
model.add(Dense(128, activation='relu', input_dim=X_train_scaled.shape[1]))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(len(set(y_pesticide)), activation='softmax'))  # Output layer

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(X_train_scaled, y_train, epochs=100, batch_size=16, validation_split=0.2)

#  Evaluate the model
loss, accuracy = model.evaluate(X_test_scaled, y_test)
print(f"\n Accuracy for 'Recommended Pesticide' prediction: {accuracy*100:.2f}%")
```
